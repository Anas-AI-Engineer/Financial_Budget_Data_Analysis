```python

import numpy as np
import pandas as pd

# CSV file ko read karke DataFrame mein store kar rahe hain
pd = pd.read_csv('Financial_Budget_Analysis.csv')
print(pd.to_string())

# DataFrame se zaroori columns alag variables mein nikal rahe hain
categories = pd["Category"]
month = pd["Month"]
budget = pd["Budgeted_Amount"]
expenses = pd["Actual_Amount"]
income = pd["Monthly_Income"]
payment = pd["Payment_Method"]
expense_type = pd["Type"]
essentail = pd["Essential"]


# 1. General Statistics
# Total, average, median, highest, lowest aur expenses ka spread check karte hain
def general_statistics():
    print("===== GENERAL STATISTICS =====")
    print("Total Transaction:", len(expenses))
    print("Total Expense:", round(np.sum(expenses), 2))
    print("Average Expense:", round(np.mean(expenses), 2))
    print("Median Expense:", round(np.median(expenses), 2))
    print("Highest Expense:", round(np.max(expenses), 2))
    print("Lowest Expense:", round(np.min(expenses), 2))
    print("Standard Deviation:", round(np.std(expenses), 2))


# 2. Category Analysis
# Har category ke total aur average expenses calculate karte hain
def category_analysis():
    print("==== CATEGORY WISE ANALYSIS ====")
    print("=" * 40)

    # Categories ki unique values nikalte hain
    unique_categories = np.unique(categories)

    for category in unique_categories:
        # Mask sirf current category wali rows select karta hai
        mask = categories == category
        total = np.sum(expenses[mask])
        average = np.mean(expenses[mask])

        print(
            category,
            "| Total:", round(total, 2),
            "| Average:", round(average, 2)
        )


# 3. Top 10 Expenses Analysis
# Sab se zyada expenses wali 10 transactions find karte hain
def top_expenses():
    print("=" * 40)
    print("====== TOP 10 EXPENSES ======")

    # argsort expenses ke sorted indices deta hai;
    # [::-1] unhein descending order mein reverse karta hai
    sorted_indices = np.argsort(expenses)[::-1]
    top_indices = sorted_indices[:10]

    for index in top_indices:
        print(
            categories[index],
            "| Amount:", round(expenses[index], 2),
            "| Month:", month[index]
        )


# 4. Budget Comparison
# Actual expenses ko budget se compare karke spending status check karte hain
def budget_analysis():
    print("=" * 40)
    print("==== BUDGET COMPARISON ====")

    # Positive difference ka matlab budget se zyada kharcha
    difference = expenses - budget

    over_spending = difference > 0
    under_spending = difference < 0
    within_budget = difference == 0

    print("Over Spending Budget:", np.sum(over_spending))
    print("Under Spending Budget:", np.sum(under_spending))
    print("Within Budget:", np.sum(within_budget))

    print("Total Budgeted Amount:", round(np.sum(budget), 2))
    print("Total Difference:", round(np.sum(difference), 2))
    print("Total Actual Expense:", round(np.sum(expenses), 2))


# 5. Monthly Expense Analysis
# Har month ke tamam expenses ka total calculate karte hain
def monthly_analysis():
    print("=" * 40)
    print("==== MONTHLY EXPENSE ANALYSIS ====")

    unique_months = np.unique(month)

    for m in unique_months:
        # Current month ki transactions select karte hain
        mask = month == m
        total = np.sum(expenses[mask])
        print("Month:", m, "| Total:", round(total, 2))


# 6. Income and Expense Analysis
# Total income aur expenses compare karke expense-to-income ratio nikalte hain
def income_analysis():
    print("=" * 40)
    print("==== INCOME AND EXPENSE ANALYSIS ====")

    total_income = np.sum(income)
    total_expense = np.sum(expenses)

    print("Total Income:", round(total_income, 2))
    print("Total Expenses:", round(total_expense, 2))
    print("Average Income:", round(np.mean(income), 2))

    # Zero income hone par division se bachne ke liye check
    if total_income > 0:
        ratio = (total_expense / total_income) * 100
        print("Expense to Income Ratio:", round(ratio, 2))


# 7. Payment Method Analysis
# Har payment method ki transaction count aur total spending check karte hain
def payment_analysis():
    print("=" * 40)
    print("==== PAYMENT METHOD ANALYSIS ====")

    unique_method = np.unique(payment)

    for method in unique_method:
        mask = payment == method
        total = np.sum(expenses[mask])
        count = np.sum(mask)

        print(
            method,
            "| Transactions:", count,
            "| Total:", round(total, 2)
        )


# 8. Expense Type Analysis
# Needs, wants ya savings jaisi types ke hisaab se expenses analyze karte hain
def type_analysis():
    print("=" * 40)
    print("==== EXPENSE TYPE ANALYSIS ====")

    unique_type = np.unique(expense_type)

    for item in unique_type:
        mask = expense_type == item
        print(
            item,
            "| Count:", np.sum(mask),
            "| Total:", round(np.sum(expenses[mask]), 2)
        )


# 9. Essential vs Non-Essential Analysis
# Essential aur non-essential expenses ki count aur total nikalte hain
def essential_analysis():
    print("=" * 40)
    print("==== ESSENTIAL EXPENSE ANALYSIS ====")

    for value in np.unique(essentail):
        mask = essentail == value
        print(
            value,
            "| Count:", np.sum(mask),
            "| Total:", round(np.sum(expenses[mask]), 2)
        )


# 10. Correlation Analysis
# Income aur expenses ke darmiyan relationship check karte hain
def correlation_analysis():
    print("=" * 40)

    # Correlation ke liye dono columns mein variation aur enough data hona chahiye
    if len(expenses) > 1 and np.std(expenses) > 0 and np.std(income) > 0:
        correlation = np.corrcoef(expenses, income)[0, 1]
        print("Income vs Expenses Correlation:", round(correlation, 3))
    else:
        print("Correlation calculate nahi ho sakti bhai.")


# 11. Search Category
# User ki di hui category ko search karke uske expenses analyze karte hain
def search_category():
    print("=" * 40)
    print("==== SEARCH CATEGORY ====")

    # strip extra spaces hatata hai aur lower case se search consistent hoti hai
    search = input("Category Enter Karo: ").strip().lower()

    # Har category ko normalize karke user ke input se compare karte hain
    mask = np.array([
        str(c).strip().lower() for c in categories
    ]) == search

    # User ko available categories dikhate hain
    print("Available Categories:", categories.unique())

    if np.any(mask):
        print("Total Transactions:", np.sum(mask))
        print("Total Expenses:", round(np.sum(expenses[mask]), 2))
        print("Average Expenses:", round(np.mean(expenses[mask]), 2))
    else:
        print("Invalid category!")


# Main Menu
# User ko options deta hai taake woh apni marzi ka analysis chala sake
while True:
    print("\n===== FINANCIAL BUDGET ANALYSIS =====")
    print("1. General Statistics")
    print("2. Category Analysis")
    print("3. Top 10 Expenses")
    print("4. Budget Comparison")
    print("5. Monthly Analysis")
    print("6. Income Analysis")
    print("7. Payment Method Analysis")
    print("8. Needs, Wants and Savings")
    print("9. Essential Expense Analysis")
    print("10. Correlation Analysis")
    print("11. Search Category")
    print("0. Exit")

    # User ki choice ko string ke taur par lete hain
    choice = input("Apni choice enter karo: ").strip()

    # Choice ke mutabiq relevant function call hota hai
    if choice == "1":
        general_statistics()
    elif choice == "2":
        category_analysis()
    elif choice == "3":
        top_expenses()
    elif choice == "4":
        budget_analysis()
    elif choice == "5":
        monthly_analysis()
    elif choice == "6":
        income_analysis()
    elif choice == "7":
        payment_analysis()
    elif choice == "8":
        type_analysis()
    elif choice == "9":
        essential_analysis()
    elif choice == "10":
        correlation_analysis()
    elif choice == "11":
        search_category()
    elif choice == "0":
        print("Financial Analysis complete. Allah Hafiz!")
        break
    else:
        # Agar choice valid na ho toh dobara try karne ko kehte hain
        print("Invalid choice! Dobara try karo.")
```
